```python
import json
import urllib.request

url = "https://api.polygon.io/v3/reference/tickers?market=fx&active=true&order=asc&limit=100&sort=ticker&apiKey=ySVTqEYU5OoF_j9Ic5mufeVhvvMFyhLt"

with urllib.request.urlopen(url) as response:
    data = json.loads(response.read().decode())

with open('tickers.txt', 'w') as f:
    for item in data['results']:
        f.write(f"{item['ticker']}\t{item['name']}\n")

print("Text file 'tickers.txt' created with name and ticker.")

with open('tickers.txt', 'r') as f:
    print(f.read())
```
